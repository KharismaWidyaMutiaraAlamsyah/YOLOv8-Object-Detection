```python
#  melatih model deteksi objek YOLO menggunakan modul ultralytics
from ultralytics import YOLO

model = YOLO("yolov8x.pt") # load a pretrained model

# Train the model
results = model.train(data='/content/drive/MyDrive/yolov8/data.yaml',
                      name='jajanan', epochs=100, patience=20, batch=16,
                      cache=True, imgsz=640, iou=0.5,
                      augment=True, degrees=25.0, fliplr=0.0, lr0=0.0001, optimizer='Adam')
```

```python
def estimate_tp_fn_fp(results):
    total_instances = results['all']['Instances']
    total_detected_instances = sum(results[class_name]['Instances'] for class_name in results.keys() if class_name != 'all')

    tp = total_detected_instances  # True Positive adalah selisih antara total_detected_instances dan FN
    fn = max(0, total_instances - tp)  # Memastikan FN tidak negatif
    fp = max(0, total_detected_instances - total_instances)  # Memastikan FP tidak negatif

    return tp, fn, fp

def calculate_accuracy(tp, fp, fn):
    # Menghitung akurasi berdasarkan estimasi TP, FP, dan FN
    accuracy = tp / (tp + fp + fn)  # Akurasi = TP / (TP + FP + FN)
    return accuracy

# Contoh data dari hasil validasi yang diberikan
results = {
    'all': {'Instances': 127},
    'Amparan_Tatak': {'Instances': 4},
    'Getuk_Pelangi': {'Instances': 5},
    'Kue_Bikaambon': {'Instances': 5},
    'Kue_Klepon': {'Instances': 21},
    'Kue_Putuayu': {'Instances': 43},
    'Lapis_Pandan': {'Instances': 4},
    'Lemper': {'Instances': 5},
    'Risoles': {'Instances': 5},
    'kue_serabi': {'Instances': 14},
    'kue_palm': {'Instances': 11}
}

# Menghitung estimasi TP, FN, dan FP
tp, fn, fp = estimate_tp_fn_fp(results)

# Menghitung akurasi
accuracy = calculate_accuracy(tp, fp, fn)

print("Estimated True Positive (TP):", tp)
print("Estimated False Positive (FP):", fp)
print("Estimated False Negative (FN):", fn)
print("Accuracy:", accuracy)
```

```python
# Untuk menghitung hasil total Accuracy, Precision, dan Recall
total_accuracy = (accuracy)
precision = tp/(tp+fp)
recall = tp/(tp+fn)
f1_score = (2*precision*recall/(precision+recall))
print("Accuracy:", total_accuracy,'%')
print("precision:", precision,'%')
print("recall:", recall,'%')
print("f1_score:", f1_score,'%')
```

```python
import matplotlib.pyplot as plt

# Membuat grafik batang
metrics = ['Accuracy', 'Precision', 'Recall', 'F1 Score']
values = [total_accuracy*100, precision*100, recall*100, f1_score*100]

plt.figure(figsize=(10, 6))
plt.bar(metrics, values, color=['blue', 'green', 'red', 'purple'])

# Menambahkan judul dan label sumbu
plt.title('Grafik Yolov8')
plt.xlabel('Metrics')
plt.ylabel('Percentage')

# Menambahkan nilai di atas batang
for i, value in enumerate(values):
    plt.text(i, value + 1, f'{value:.2f}%', ha='center', va='bottom')

plt.ylim(0, 110)  # Batas sumbu y dari 0 hingga 110 untuk tampilan yang lebih baik
plt.show()
```

```python
# Import library yang dibutuhkan
from PIL import Image
import cv2
from google.colab.patches import cv2_imshow
from ultralytics import YOLO

# Memuat model YOLO
model = YOLO('./runs/detect/jajanan/weights/best.pt')

# Melakukan deteksi pada gambar
deteksi = model('/content/nastarkastangel.jpeg')

# Menyimpan gambar yang telah dideteksi
for r in deteksi:
    im_array = r.plot()  # plot a BGR numpy array of predictions
    im = Image.fromarray(im_array[..., ::-1])  # RGB PIL image
    img = cv2.cvtColor(im_array[..., ::-1], cv2.COLOR_BGR2RGB)  # Open-CV reads images as BGR, we convert it to RG

    # Menyimpan gambar hasil deteksi
    output_path = '/content/drive/MyDrive/hasil/hasil.jpg'
    cv2.imwrite(output_path, img)

    detected_classes = set()  # Menggunakan set untuk mengumpulkan nama kelas unik

    # Mengambil semua deteksi dari hasil model
    for det in r.boxes.data:
        class_id = int(det[5])  # indeks kelas
        class_name = r.names[class_id]  # nama kelas berdasarkan indeks

        # Menambahkan nama kelas ke set
        detected_classes.add(class_name)

    # Menampilkan nama-nama kelas yang terdeteksi
    print("Hasil Deteksi :", ", ".join(detected_classes))
    value = "".join(detected_classes)
    cv2_imshow(img)

# Menginputkan pecarian ke gemini
model = genai.GenerativeModel('gemini-pro')
respon = model.generate_content( "Resep dan langkah"+value)
to_markdown(respon.text)
```
